### Checking a hand-derived gradient against PyTorch's autograd

`lst-dl-backprop-check`

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
x_ = torch.tensor(2.0)
target = torch.tensor(5.0)

y_hat = w * x_ + b
loss = (y_hat - target) ** 2
loss.backward()
w.grad, b.grad
```

### A minimal training loop: repeated gradient updates on synthetic data

`lst-dl-training-loop`

```python
torch.manual_seed(0)
x = torch.randn(32, 10)
y = torch.randn(32, 1)

model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),
    nn.Linear(64, 1),
)
loss_fn = nn.MSELoss()
opt = torch.optim.Adam(model.parameters(), lr=1e-2)

for step in range(8):
    opt.zero_grad()
    pred = model(x)
    loss = loss_fn(pred, y)
    loss.backward()
    opt.step()
    print(f"step {step}: loss = {loss.item():.4f}")
```
